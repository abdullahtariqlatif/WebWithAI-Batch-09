```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Filter down to main repository records
repos_df = df[df["stars"].notna() & df["created_at"].notna()].copy()

# Convert dates and calculate year & age in years
repos_df["created_at"] = pd.to_datetime(repos_df["created_at"], errors="coerce")
repos_df["created_year"] = repos_df["created_at"].dt.year

# Use 2026 as reference year since dataset is from 2026
repos_df["age_years"] = 2026 - repos_df["created_year"]

print(f"Cleaned repository records: {len(repos_df):,} rows")
```

```python
# Focus on Ruby and major comparison languages
target_langs = ["Ruby", "Python", "JavaScript", "Go", "Java"]
filtered_langs = repos_df[repos_df["primary_language"].isin(target_langs)]

# Group by creation year and language
yearly_trends = (
    filtered_langs.groupby(["created_year", "primary_language"])
    .size()
    .reset_index(name="repo_count")
)

# Plot Line Chart
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

sns.lineplot(
    data=yearly_trends,
    x="created_year",
    y="repo_count",
    hue="primary_language",
    marker="o",
    linewidth=2.5,
)

plt.title("Popularity Trends Over Time: Ruby vs. Major Languages", fontsize=14)
plt.xlabel("Year Created", fontsize=12)
plt.ylabel("Number of Repositories", fontsize=12)
plt.legend(title="Language", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Categorize repository types
repos_df["repo_category"] = repos_df["primary_language"].apply(
    lambda x: (
        "Non-Programming / General"
        if pd.isna(x) or str(x).strip() in ["", "None"]
        else "Language-Based"
    )
)

# Plot Scatter Chart
plt.figure(figsize=(12, 6))

sns.scatterplot(
    data=repos_df,
    x="age_years",
    y="stars",
    hue="repo_category",
    alpha=0.7,
    s=70,
    palette={"Language-Based": "#2ecc71", "Non-Programming / General": "#e74c3c"},
)

# Use logarithmic scale to handle star count outliers gracefully
plt.yscale("log")

plt.title(
    "Star Distribution: Language-Based vs. Non-Programming Repositories",
    fontsize=14,
)
plt.xlabel("Repository Age (Years)", fontsize=12)
plt.ylabel("Stars (Log Scale)", fontsize=12)
plt.legend(title="Category")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# 1. Filter for valid repository rows
repos_df = df[df["stars"].notna() & df["created_at"].notna()].copy()

# 2. Define repository categories (Language-Based vs Non-Programming)
repos_df["repo_category"] = repos_df["primary_language"].apply(
    lambda x: (
        "Non-Programming"
        if pd.isna(x) or str(x).strip() in ["", "None"]
        else "Language-Based"
    )
)

# 3. Select key numerical metrics
metrics = ["stars", "forks", "open_issues", "size_kb"]

# Filter out rows with missing metric values
pairplot_data = repos_df[metrics + ["repo_category"]].dropna().copy()

# Apply log1p transformation (log(x + 1)) to handle heavy skewness/outliers
log_metrics = [f"log_{m}" for m in metrics]
for m in metrics:
    pairplot_data[f"log_{m}"] = np.log1p(pairplot_data[m])

# 4. Generate Pair Plot
sns.set_theme(style="ticks")

g = sns.pairplot(
    data=pairplot_data,
    vars=log_metrics,
    hue="repo_category",
    diag_kind="kde",  # Kernel Density Estimate on the diagonal
    corner=True,  # Plot lower triangle only for cleaner layout
    palette={"Language-Based": "#2ecc71", "Non-Programming": "#e74c3c"},
    plot_kws={"alpha": 0.5, "s": 25},
)

g.fig.suptitle(
    "Pairwise Relationships Between Repository Metrics (Log Scale)", y=1.02
)
plt.show()
```
