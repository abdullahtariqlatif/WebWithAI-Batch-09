```python
import matplotlib.pyplot as plt
import numpy as np

# Data
xpoints = np.array([0, 6])
ypoints = np.array([0, 250])

# Plotting
plt.plot(xpoints, ypoints)
plt.title("Basic Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

# Display (Required for .py scripts; optional in .ipynb)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 6, 8])
y = np.array([3, 8, 1, 10])

# 'o:r' means circles ('o'), dotted line (':'), red color ('r')
plt.plot(x, y, 'o:r', ms=10, mfc='blue')
plt.title("Customized Line and Markers")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Bar Chart
x_bars = np.array(["A", "B", "C", "D"])
y_bars = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 1)  # 1 row, 2 columns, chart index 1
plt.bar(x_bars, y_bars, color="#4CAF50")
plt.title("Bar Chart")

# Scatter Plot
x_scatter = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y_scatter = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])

plt.subplot(1, 2, 2)  # 1 row, 2 columns, chart index 2
plt.scatter(x_scatter, y_scatter, color="hotpink")
plt.title("Scatter Plot")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load built-in sample dataset
fmri = sns.load_dataset("fmri")

# Line plot showing confidence intervals automatically
sns.lineplot(data=fmri, x="timepoint", y="signal", hue="event", style="event")
plt.title("Seaborn Line Plot with CI")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load sample dataset
penguins = sns.load_dataset("penguins")

# Histogram with Kernel Density Estimate (KDE) curve
sns.displot(penguins, x="flipper_length_mm", hue="species", kde=True)
plt.title("Penguin Flipper Length Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

# Box plot to analyze statistical spread across categories
sns.boxplot(x="day", y="total_bill", hue="smoker", data=tips, palette="Set3")
plt.title("Total Bill Distribution by Day and Smoking Status")
plt.show()
```
