```python
import pandas as pd

# Load dataset
df = pd.read_csv("DataSets/Strength Progression: 5.8M Sets, 10,407 Lifters/strength-report-2026.csv")

# 1. Extract the 'milestones' table
milestones_df = df[df['dataset'] == 'milestones']

# 2. Pivot metrics into separate columns for analysis
milestones_pivoted = milestones_df.pivot(
    index=['key1', 'key2', 'key3'],
    columns='metric',
    values='value'
).reset_index()

# Rename keys for clarity
milestones_pivoted.columns.name = None
milestones_pivoted.rename(columns={
    'key1': 'exercise',
    'key2': 'milestone_tier',
    'key3': 'months_logged'
}, inplace=True)

# Preview clean tabular format
print(milestones_pivoted.head())
```

```python
import pandas as pd

df = pd.read_csv("DataSets/Strength Progression: 5.8M Sets, 10,407 Lifters/strength-report-2026.csv")

def get_clean_table(dataset_name, key_names):
    """
    Extracts a sub-table, pivots metrics into columns, and renames keys.
    """
    # 1. Filter dataset
    sub_df = df[df['dataset'] == dataset_name]

    # 2. Pivot metrics to columns
    pivoted = sub_df.pivot(
        index=['key1', 'key2', 'key3'],
        columns='metric',
        values='value'
    ).reset_index()

    pivoted.columns.name = None

    # 3. Rename active keys and drop unused ones
    rename_dict = {}
    for i, name in enumerate(key_names, start=1):
        if name:
            rename_dict[f'key{i}'] = name

    pivoted = pivoted.rename(columns=rename_dict)

    # Drop leftover key columns that weren't assigned a name
    unused_keys = [f'key{i}' for i in range(len(key_names) + 1, 4)]
    pivoted = pivoted.drop(columns=[k for k in unused_keys if k in pivoted.columns])

    return pivoted

# Example: Get top exercises table cleaned up directly
top_exercises = get_clean_table('top-exercises', ['exercise_name', 'type'])
print(top_exercises.head())
```
